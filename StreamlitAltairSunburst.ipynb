```python
%%writefile test.py
from pathlib import Path

import streamlit as st
import plotly.express as px
import pandas as pd
import numpy as np
import plotly.graph_objects as go
import altair as alt

DATA_PATH = Path(__file__).resolve().parent / "SGJob_ITSalaryClean.csv"

@st.cache_data
def load_data():
    return pd.read_csv(
        DATA_PATH,
        parse_dates=[
            "metadata_newPostingDate",
            "metadata_originalPostingDate",
            "metadata_expiryDate",
    ],
)

df = load_data()

chart_col1, chart_col2 = st.columns(2)

with chart_col1:
    st.subheader("📈 Altair chart")
    chart = alt.Chart(df).mark_circle().encode(
        x='category',
        y='average_salary_clean',
        color='positionLevels',
    ).interactive()
    st.altair_chart(chart, theme="streamlit", use_container_width=True, height="stretch")

with chart_col2:
    st.subheader("📈 Sunburst chart")
    df = pd.DataFrame(
        dict(category=df['category'], company=df['postedCompany_name'], status=df['status_jobStatus'], salary=df['average_salary_clean'], experience=df['minimumYearsExperience'], positions=df['positionLevels'], employment=df['employmentTypes'])
    )
    fig = px.sunburst(df, path=['category', 'positions','employment'], color_discrete_sequence=px.colors.qualitative.Pastel)
    fig.update_layout(margin=dict(l=10, r=10, t=5, b=10))
    st.plotly_chart(fig, use_container_width=True)
```
