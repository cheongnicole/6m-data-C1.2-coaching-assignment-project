```python
import pandas as pd
import numpy as np

# Reading Jobs Data csv with exploded category column to reduce running time
df = pd.read_csv("SGJobData_exploded.csv")

# Inital filtering for IT jobs based on the category column
df_it_cat = df[df["category"] == "Information Technology"]

# Filtering for IT jobs based on keywords in the title column

# 1. Filter for keywords that are case insensitive
keywords_casefalse = ["Information Technology", "Software", "Developer", "Programmer", "System", "Network", "Database", "Administrator", "Cloud", "Infrastructure", "DevOps", "Machine Learning", "Python", "Java", "C#", "JavaScript", "HTML", "Linux", "Unix", "Cybersecurity", "Data Science", "Business Intelligence", "Power BI", "Oracle", "Azure", "Scrum", "Agile", "Murex", "Technical Support","Chatbot","Algo"]
# 2. Filter for keywords that are case sensitive
keywords_casetrue = ["IT", "ML", "CSS", "SQL", "SAP","GCP"]
df_it_keywords_casefalse = df[df["title"].str.strip().str.contains("|".join(keywords_casefalse), case=False, na=False)]
df_it_keywords_casetrue = df[df["title"].str.strip().str.contains("|".join(keywords_casetrue), case=True, na=False)]

# 3. Combine the two filtered dataframes and drop duplicates
df_it_keywords = pd.merge(df_it_keywords_casefalse, df_it_keywords_casetrue, how='outer').drop_duplicates()

# 4. Filter by negative keywords in the title column, esp. words containing "it" but not IT related
negative_keywords = [
    "Archit","Audit", "Nutrit", "Edit", "Writ", "Janit","Blacksmith", "Locksmith", "Goldsmith", "Silversmith", "Wait",
    "Recruit", "Solicit", "Credit", "Equit", "Exhibit", "Kitchen", "Curriculum", "Sales", "Quantity", "Quality", "ASAP", "Position", "Site", "City", "Nitec", "Security Officer","Security Guard","Instit","Fire","HIIT","Human Resource","Hospit","Teacher","Chef","Food","Benefit","Health","Pharma","Customer","Lecturer","HR", "School", "Clinic", "Chemist", "Compliance", "Furnit", "Music", "Education", "Funeral","Fitness"
]    
df_it_keywords = df_it_keywords[~df_it_keywords["title"].str.contains("|".join(negative_keywords), case=False, na=False)]

# 5. Drop incorrect categories by id
invalid_ids = [2, 6, 8, 10, 15, 18, 19, 20, 28, 32, 33]
reverse_mask = df_it_keywords["id"].isin(invalid_ids)
mask = ~reverse_mask
df_it_keywords = df_it_keywords[mask]

# 5. Combine the two filtered dataframes and drop duplicates
df_it = pd.merge(df_it_cat, df_it_keywords, how='outer').drop_duplicates()

# 6. Write the filtered IT jobs dataframe to a new CSV file
df_it.to_csv("SGJobData_IT.csv", index=False)

```
