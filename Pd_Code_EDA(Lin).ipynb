# DUPLICATES

```python
# Checking for Duplicates

df.duplicated().sum()
#returns 3987

df.duplicated(subset='metadata_jobPostId', keep = False).sum()
#returns 3988
# Jobpost ID should be unique 

df['metadata_jobPostId'].isna().sum()
#returns 3988
#These jobpost ID entries may not be duplicates, they are Null 

# Check within the subset, how many duplicates
mask1 = df[df.duplicated(subset='metadata_jobPostId', keep=False)]
# print(sub)
mask1.isna().sum(axis=1) #within each row, 12 are null
mask1.isna().sum(axis=1).value_counts() #returns 3988

# Comparing to the above missing value portion, these 3988 entries are all null
# Therefore we should drop these rows
```
