```python
import pandas as pd
import numpy as np
import json

# Create dateframe df using jobs data
df = pd.read_csv("SGJobData.csv")

# Convert JSON strings back into Python
df['categories'] = df['categories'].fillna('{}').apply(json.loads)

# Explode the categories column to separate rows
df_exploded = df.explode('categories').reset_index(drop=True)

# Normalize the dictionary column to create new id and category columns
df_exploded= df_exploded.drop(columns=['categories']).join(pd.json_normalize(df_exploded['categories']))

# Write the exploded dataframe to a new CSV file for use in other parts of code to reduce running time
df_exploded.to_csv("SGJobData_exploded.csv")
```
