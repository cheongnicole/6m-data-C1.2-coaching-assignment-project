#### 1.3. Investigate if `categories` JSON is clean

```python
# Parse the JSON in the "categories" column into a list of dictionaries
df["categories_parsed"] = df["categories"].apply(json.loads)

# Are there any rows with empty categories?
empty_categories = df[df["categories_parsed"].apply(lambda x: len(x) == 0)]
print("Rows with empty categories:")
print(empty_categories.shape[0])

# Are there any rows with missing categories?
missing_categories = df[df["categories_parsed"].isnull()]
print("Rows with missing categories:")
print(missing_categories.shape[0])

# Are there any rows with duplicate categories? e.g. [{"category": "Software Engineer"}, {"category": "Software Engineer"}]
def has_duplicate_categories(categories):
    seen = set()
    for category in categories:
        cat_name = category.get("category")
        if cat_name in seen:
            return True
        seen.add(cat_name)
    return False
duplicate_categories = df[df["categories_parsed"].apply(has_duplicate_categories)]
print("Rows with duplicate categories:")
print(duplicate_categories.shape[0])
```

#### 1.4. Explode `categories` JSON

```python
# Explode the "categories_parsed" column so that each category gets its own row
df_exploded = df.explode("categories_parsed")
# create new columns for category_id and category_name
df_exploded["category_id"] = df_exploded["categories_parsed"].apply(lambda x: x.get("id") if isinstance(x, dict) else None)
df_exploded["category_name"] = df_exploded["categories_parsed"].apply(lambda x: x.get("category") if isinstance(x, dict) else None)
df_exploded.head(3)

# List all unique categories sort by category_id
all_categories = df_exploded[["category_id", "category_name"]].drop_duplicates().sort_values(by="category_id")
all_categories

# Count the number of occurrences of each category
category_counts = df_exploded["category_name"].value_counts()
category_counts
```

### 2. Visualization with Streamlit

```python
%%writefile assignment.py
from pathlib import Path

import pandas as pd
import streamlit as st

DATA_PATH = Path(__file__).resolve().parent / "SGJobData_cleaned.csv"
df = pd.read_csv(
    DATA_PATH,
    parse_dates=[
        "metadata_newPostingDate",
        "metadata_originalPostingDate",
        "metadata_expiryDate",
    ],
)

EXPLODED_DATA_PATH = Path(__file__).resolve().parent / "SGJobData_exploded.csv"
df_exploded = pd.read_csv(
    EXPLODED_DATA_PATH,
    parse_dates=[
        "metadata_newPostingDate",
        "metadata_originalPostingDate",
        "metadata_expiryDate",
    ],
)

st.title("Singapore Job Market Dashboard")

with st.sidebar:
    st.header("Filters")
    category_options = sorted(df_exploded["category_name"].dropna().unique())
    selected_categories = st.multiselect(
        "Category", category_options, default=category_options
    )

    employment_options = sorted(df["employmentTypes"].dropna().unique())
    selected_employment = st.selectbox("Employment Type", ["All"] + employment_options)

    open_only = st.checkbox("Show only open postings")

    min_exp = int(df["minimumYearsExperience"].min())
    max_exp = int(df["minimumYearsExperience"].max())
    experience_range = st.slider(
        "Minimum Years Experience", min_exp, max_exp, (min_exp, max_exp)
    )

# df = df[df["primary_category"].isin(selected_categories)]
if selected_employment != "All":
    df = df[df["employmentTypes"] == selected_employment]
if open_only:
    df = df[df["status_jobStatus"] == "Open"]
df = df[df["minimumYearsExperience"].between(*experience_range)]

st.header("Overview")

col1, col2, col3 = st.columns(3)
col1.metric("Total Postings", len(df))
col2.metric(
    "Average Salary",
    f"${df['average_salary'].mean():,.0f}" if len(df) else "N/A",
)
col3.metric("Total Applications", int(df["metadata_totalNumberJobApplication"].sum()))

with st.expander("View raw data"):
    st.dataframe(df)

st.header("Trends & Breakdown")

postings_by_month = (
    df.groupby(df["metadata_newPostingDate"].dt.to_period("M")).size().rename("postings")
)
postings_by_month.index = postings_by_month.index.to_timestamp()

# avg_salary_by_category = (
    # df.groupby("primary_category")["average_salary"].mean().sort_values(ascending=False)
# )

tab1, tab2, tab3 = st.tabs(["Trends", "Categories", "Salary"])

with tab1:
    st.subheader("Postings Over Time")
    st.line_chart(postings_by_month)
    st.subheader("Cumulative Postings")
    st.area_chart(postings_by_month.cumsum())

# with tab2:
    # st.subheader("Average Salary by Category")
    # st.bar_chart(avg_salary_by_category)

with tab3:
    st.subheader("Experience vs Salary")
    st.scatter_chart(df, x="minimumYearsExperience", y="average_salary")
```
