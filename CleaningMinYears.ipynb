```python
import pandas as pd
import numpy as np

# Reading IT Jobs Data csv to reduce runtime
df_it= pd.read_csv("SGJobData_IT.csv")

# Filter outliers for minimumYearsExperience column

# 1. Split into 3 groups: inexperienced (<3 years), mid-level (3-20 years inclusive of both ends), experienced (>20 years)
df_inexperienced = df_it[df_it["minimumYearsExperience"] < 3]
df_midlevel = df_it[(df_it["minimumYearsExperience"] >= 3) & (df_it["minimumYearsExperience"] <= 20)]
df_experienced = df_it[df_it["minimumYearsExperience"] > 20]

# 2. Filter for inexperienced and experienced groups

# Impossible for inexperienced group to be "Senior Management", either drop rows or modify to "Senior Executive" or "Executive".
# Set to NULL for now, but can modify to "Senior Executive" or "Executive" if needed.
df_inexperienced.loc[(df_inexperienced["positionLevels"] == "Senior Management"),"positionLevels"] = np.nan

# Drop approx 1 row from experienced group as it is impossible for experienced group to be "Junior Executive"
df_experienced = df_experienced[df_experienced["positionLevels"] != "Junior Executive"]

# Groupby positionLevels to see where the outliers are -- "Manager" and "Professional" have outliers
df_experienced.groupby("positionLevels")["minimumYearsExperience"].mean()

# Filter for "Manager", 2 outliers 56 and 60, safe to drop
#df_experienced[df_experienced["positionLevels"] == "Manager"]["minimumYearsExperience"].describe()
df_experienced = df_experienced[df_experienced["positionLevels"] != "Manager"]

# Filter for "Professional" and > 22 years of experience, 2 outliers 61 and 87, safe to drop
df_experienced[(df_experienced["positionLevels"] == "Professional")&(df_experienced["minimumYearsExperience"] > 22)]["minimumYearsExperience"].describe()
df_experienced = df_experienced[(df_experienced["positionLevels"] != "Professional")|(df_experienced["minimumYearsExperience"] <= 22)]

# Check that correct number of rows are dropped by comparing combined 3 dfs vs original df
print(len(df_experienced)+len(df_inexperienced)+len(df_midlevel), len(df_it))

# Re-merge the 3 groups back into df_it
df_it = pd.merge(pd.merge(df_inexperienced, df_midlevel, how='outer'), df_experienced, how='outer')
```
